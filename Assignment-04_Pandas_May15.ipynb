### **Problem 4: Numpy Array Operations**  
1. Convert the numerical columns (such as `LotArea`, `YearBuilt`, `GrLivArea`, `SalePrice`, etc.) into a **NumPy array**.
2. Perform the following NumPy operations on the array:
   - Compute the **mean** and **standard deviation** of each column.
   - Normalize the data using **min-max scaling** and **z-score normalization**.
   - Compute the **dot product** between two numerical columns of your choice.

```python
# convert the numerical columns to numeric data types
# Select numerical columns
numerical_cols = ['LotArea', 'YearBuilt', 'GrLivArea', 'SalePrice']

# Convert to NumPy array
data_array = house_data[numerical_cols].to_numpy()

# Display NumPy array
print("NumPy Array:\n")
print(data_array)
```

```python
house_data.info()
```

```python

import numpy as np
mean_values = np.mean(data_array, axis=0)
std_values = np.std(data_array, axis=0)

print("\nMean of each column:")
for col, mean in zip(numerical_cols, mean_values):
    print(f"{col}: {mean}")

print("\nStandard Deviation of each column:")
for col, std in zip(numerical_cols, std_values):
    print(f"{col}: {std}")
```

```python
min_values = np.min(data_array, axis=0)
max_values = np.max(data_array, axis=0)

min_max_scaled = (data_array - min_values) / (max_values - min_values)

print("\nMin-Max Scaled Data:\n")
print(min_max_scaled)
```

```python
z_score_normalized = (data_array - mean_values) / std_values

print("\nZ-Score Normalized Data:\n")
print(z_score_normalized)
```

```python
lotarea = house_data['LotArea'].to_numpy()
saleprice = house_data['SalePrice'].to_numpy()

dot_product = np.dot(lotarea, saleprice)

print("\nDot Product of LotArea and SalePrice:")
print(dot_product)
```

### **Problem 6: Aggregation and Transformation**  
1. Perform a **groupby** operation on `Neighborhood` and calculate the **mean** and **median** for `SalePrice`.
2. Use the `apply()` function to create a custom transformation for `SalePrice` (e.g., scaling or transformation).
3. Aggregate multiple statistical values (e.g., **mean**, **median**, and **count**) for `SalePrice` grouped by `OverallQual`.

```python
neighborhood_stats = house_data.groupby('Neighborhood')['SalePrice'].agg(['mean', 'median'])

print("Mean and Median SalePrice by Neighborhood:\n")
print(neighborhood_stats)
```

```python
house_data['Log_SalePrice'] = house_data['SalePrice'].apply(lambda x: np.log(x))

print("\nTransformed SalePrice (Log Scale):\n")
print(house_data[['SalePrice', 'Log_SalePrice']].head())
```

```python
min_price = house_data['SalePrice'].min()
max_price = house_data['SalePrice'].max()

house_data['Scaled_SalePrice'] = house_data['SalePrice'].apply(
    lambda x: (x - min_price) / (max_price - min_price)
)

print("\nScaled SalePrice:\n")
print(house_data[['SalePrice', 'Scaled_SalePrice']].head())
```

```python

overallqual_stats = house_data.groupby('OverallQual')['SalePrice'].agg(
    Mean='mean',
    Median='median',
    Count='count'
)

print("\nSalePrice Statistics grouped by OverallQual:\n")
print(overallqual_stats)
```
